```python
# Import fertility rate data (for age 25-29)
df_asfr = pd.read_csv('../data/cleaned/age25_29_ASFR_by_regional.csv')
display(df_asfr)
```

```python
# Import gross income data (for age 22-29)
df_gross_income_original = pd.read_csv('../data/cleaned/age22_29_five_year_yearly_gross_income_regional.csv')
display(df_gross_income_original)
```

```python
# Import affordability
df_affordability = pd.read_csv('../data/cleaned/affordability_2020_2024_regional.csv')
display(df_affordability)
```

Data cleaning to be done prior to the integration:

Gross Income Data
- Delete 'Scotland' region data as Scotland is out of the scope of our project

```python
# Delete Scotland rows in gross income data in the copied dataframe df_gross_income
df_gross_income = df_gross_income_original[(df_gross_income_original['region_name'] != 'Scotland')].copy()

display(df_gross_income)
```

```python
# Create df_all to integrate 3 dataframe
region_code_list = df_gross_income['region_code'].tolist()
region_name_list = df_gross_income['region_name'].tolist()
year_list = [2020, 2021, 2022, 2023, 2024]

# print(region_code_list)
# print(region_name_list)
# print(len(region_code_list))

# create a new empty dataframe with columns
df_all = pd.DataFrame(columns=['region_code', 'region_name', 'year'])
# add row for every 'region_code', 'region_name', 'year'
for i in range(len(region_code_list)):
    for y in year_list:
        new_row = {'region_code': region_code_list[i], 'region_name': region_name_list[i], 'year': y}
        df_all.loc[len(df_all)] = new_row

display(df_all)
```

```python
# To make sure the data being added to df_all mapping to the rows in the set of 'region_code, 'region_name' 'year' columns,
# make the rows in those colums as row index
df_all.set_index(['region_code', 'region_name', 'year'], inplace = True)
display(df_all)
```

Add new column of fertility rate data in df_all dataframe

```python
# Create a list of dictionaries for fertility rate data in each row
# To create list of dictionaries, use to_dict(orient='records')
for row_fertility_dict in df_asfr.to_dict(orient='records'):
    # print(row_income_dict)
    
    # for each row in the df_asfr table
    region_code = row_fertility_dict['region_code']
    region_name = row_fertility_dict['region_name']

    for year in year_list:

        # for each year, get the row_fertility_dict value from the row dictionary
        asfr = row_fertility_dict[f'year_{year}']
        # print(gross_income)

        # push this gross_income value, into the df_all table
        df_all.loc[(region_code, region_name, year), 'fertility_rate'] = asfr
```

```python
# Check
display(df_all)
```

Add new column of gross income data in df_all dataframe

```python
# Create a list of dictionaries for income data in each row
# To create list of dictionaries, use to_dict(orient='records')
for row_income_dict in df_gross_income.to_dict(orient='records'):
    # print(row_income_dict)
    
    # for each row in the df_gross_income table
    region_code = row_income_dict['region_code']
    region_name = row_income_dict['region_name']

    for year in year_list:

        # for each year, get the row_income_dict value from the row dictionary
        gross_income = row_income_dict[f'{year}_median_yearly_gross']
        # print(gross_income)

        # push this gross_income value, into the df_all table
        df_all.loc[(region_code, region_name, year), "median_yearly_gross"] = gross_income
```

```python
# Check
display(df_all)
```

Add new column of affordability data in df_all dataframe

```python
# Create a list of dictionaries for affordability data in each row
# To create list of dictionaries, use to_dict(orient='records')
for row_affordability_dict in df_affordability.to_dict(orient='records'):
    # print(row_income_dict)
    
    # for each row in the df_gross_income table
    region_code = row_affordability_dict['region_code']
    region_name = row_affordability_dict['region_name']

    for year in year_list:

        # for each year, get the row_income_dict value from the row dictionary
        affordability = row_affordability_dict[f'{year}']
        # print(gross_income)

        # push this gross_income value, into the df_all table
        df_all.loc[(region_code, region_name, year), "median_affordability"] = affordability
```

```python
# Check
display(df_all)
```
