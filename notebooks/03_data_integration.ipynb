Import cleaned dataset for data integration\
Imported data:
- Data 1: age25_29_ASFR_by_regional.csv
- Data 2: age22_29_five_year_yearly_gross_income_regional.csv
- Data 3: yearly_house_price_2020_2026Q1_regional.csv

```python
# Import fertility rate data (for age 25-29)
df_asfr = pd.read_csv('../data/cleaned/age25_29_ASFR_by_regional.csv')
display(df_asfr)
```

```python
# Import gross income data (for age 22-29)
df_gross_income_original = pd.read_csv('../data/cleaned/age22_29_five_year_yearly_gross_income_regional.csv')
display(df_gross_income_original)
```

```python
# Import house price
df_house_price_original = pd.read_csv('../data/cleaned/yearly_house_price_2020_2026Q1_regional.csv')
```

```python
# Create a copy of df_house_price as the layout of this dataframe is the final layout after the integration
# therefore dataframe build based on this dataframe 
df_house_price_copy = df_house_price_original.copy()

display(df_house_price_copy)
```

```python
# Change the column name from house_price to average_house_price
# check the column name changed as expected
# df_house_price_copy.rename(columns = {'house_price': 'average_house_price'})

# change the name in the dataframe
df_house_price_copy.rename(columns = {'house_price': 'average_house_price'}, inplace=True)
display(df_house_price_copy)
```

Data cleaning to be done prior to the integration:

Gross Income Data
- Delete 'Scotland' region data as Scotland is out of the scope of our project

House Price Data
- Delete year 2025 and 2026 rows as other dataset do not have the data for 2025 and 2026 (MCAR in imcome and fertility datasets)
- Delete 'Scotland' region data as Scotland is out of the scope of our project

```python
# Delete Scotland rows in gross income data in the copied dataframe df_gross_income
df_gross_income = df_gross_income_original[(df_gross_income_original['region_name'] != 'Scotland')].copy()

display(df_gross_income)
```

```python
# df_all will be used to integrate the three dataframe
# Delete year 2025, 2026 and Scotland rows in house price data
df_all = df_house_price_copy[(df_house_price_copy['year'] != 2025) & 
                             (df_house_price_copy['year'] != 2026) & 
                             (df_house_price_copy['region_name'] != 'Scotland')].copy()

# crate list of 'year' to use later
# year_list = df_all[(df_all['year'] != 2025) & (df_all['year'] != 2026)]['year'].unique().tolist()
year_list = df_all['year'].unique().tolist()
print(year_list)

display(df_all)
```

```python
# To make sure the data being added to df_all mapping to the rows in the set of 'region_code, 'region_name' 'year' columns,
# make the rows in those colums as row index
df_all.set_index(['region_code', 'region_name', 'year'], inplace = True)
display(df_all)
```

Add new column of gross income data in df_all dataframe

```python
# Create a list of dictionaries for income data in each row
# To create list of dictionaries, use to_dict(orient='records')
for row_income_dict in df_gross_income.to_dict(orient='records'):
    # print(row_income_dict)
    
    # for each row in the df_gross_income table
    region_code = row_income_dict['region_code']
    region_name = row_income_dict['region_name']

    for year in year_list:

        # for each year, get the row_income_dict value from the row dictionary
        gross_income_col_name = f'{year}_median_yearly_gross'
        gross_income = row_income_dict[gross_income_col_name]
        # print(gross_income)

        # push this gross_income value, into the df_all table
        df_all.loc[(region_code, region_name, year), "median_yearly_gross"] = gross_income
```

```python
# Check
display(df_all)
```

Repeat the same for fertility rate

```python
# To see the original dataframe
display(df_asfr)
```

```python
# Create a list of dictionaries for income data in each row
# To create list of dictionaries, use to_dict(orient='records')
for row_fertility_dict in df_asfr.to_dict(orient='records'):
    # print(row_income_dict)
    
    # for each row in the df_gross_income table
    region_code = row_fertility_dict['region_code']
    region_name = row_fertility_dict['region_name']

    for year in year_list:

        # for each year, get the row_income_dict value from the row dictionary
        fertility_col_name = f'year_{year}'
        asfr = row_fertility_dict[fertility_col_name]
        # print(gross_income)

        # push this gross_income value, into the df_all table
        df_all.loc[(region_code, region_name, year), 'fertility_rate'] = asfr
```

```python
# Check
display(df_all)
```
