##### Data Cleaning Age Specific Fertility Rate (age 25-29) #####

```python
# import Pandas

import pandas as pd
```

Note: When the csv file was created:
- Deleted unneccesary 5 rows above headers which were not impacting data.
- Changed header names adding 'year_' before year\
    example: 2019 -> year_2019
- Added column name 'region_code' where empty header was.

```python
# import nomis_ASFR_25_29.csv into Panda

data = pd.read_csv('../data/raw/modified_headers/fertility_rate/nomis_ASFR_25_29.csv')
df = pd.DataFrame(data)
display(df)
```

Note: Before cleaning the data, dropped irrelevant data as we only need regional level

```python
# Check the area name and region_code
df[['Area', 'region_code']].head(20)
region_level = df[(df['region_code'] == 'E12000001') 
                | (df['region_code'] == 'E12000002') 
                | (df['region_code'] == 'E12000003') 
                | (df['region_code'] == 'E12000004') 
                | (df['region_code'] == 'E12000005') 
                | (df['region_code'] == 'E12000006') 
                | (df['region_code'] == 'E12000007') 
                | (df['region_code'] == 'E12000008') 
                | (df['region_code'] == 'E12000009') 
                | (df['region_code'] == 'W92000004') 
                ]
display(region_level)

print('*Note: There are two Wales rows with the same data as it appears one in region level and one in country level')
```

Create the regional level only DataFrame

```python
# Copy the origianl the table to make regional level only table (10 regions)
df_copy = df.copy()
```

```python
# Delete all the rows except where the Area name starts with 'gor:'
df_by_region = df_copy[df_copy['Area'].str.startswith('gor:')].copy()
display(df_by_region)
```

```python
# Detailed information about the region DataFrame
print(f'Detailed information:\n{df_by_region.info()}')
```

```python
# Statistical summary
print('Statistical summary:')
print(f'\nyear_2019:\n{df_by_region['year_2019'].describe()}')
print(f'\nyear_2020:\n{df_by_region['year_2020'].describe()}')
print(f'\nyear_2021:\n{df_by_region['year_2021'].describe()}')
print(f'\nyear_2022:\n{df_by_region['year_2022'].describe()}')
print(f'\nyear_2023:\n{df_by_region['year_2023'].describe()}')
print(f'\nyear_2024:\n{df_by_region['year_2024'].describe()}')
```

```python
# Find missing values
print('Value Missing:')
print(df_by_region.isnull().sum())
print(f'\nTotal missing values: {df_by_region.isnull().sum().sum()}')
```

```python
# Check duplicate records
print('Duplicate Records:')
print(f'Number of duplicate rows: {df_by_region.duplicated().sum()}')
```

```python
# Find inconsistent formats
print('Inconsistetnt Formats:')
print('Area column issue')
areas = [area for area in df_by_region['Area'] if str(area).startswith(' ') or str(area).endswith(' ')]   # used str() to avoid attribute errors on NaN rows
print(f'- Whitespace issue: {len(areas)}') 
print(f'- Empty strings: {df_by_region[df_by_region['Area'] == ''].shape[0]}')

print('\nregion_code column issue:')
codes = [code for code in df_by_region['region_code'] if str(code).startswith(' ') or str(code).endswith(' ')]   # used str() to avoid attribute errors on NaN rows
print(f'- Whitespace issue in region_code column: {len(codes)}') 
print(f'- Empty strings: {df_by_region[df_by_region['region_code'] == ''].shape[0]}')
```

```python
# remove 'gor:' in Area colummn to be align with income and house_price dataset

df_by_region['Area'] = df_by_region['Area'].str.replace('gor:', '')
display(df_by_region)
```

```python
# Change to align with income and house_price dataset column names
# Change 'Area' to 'regional_name' and region_code to regional_code

df_by_region.rename(columns={'Area': 'region_name'}, inplace=True)

display(df_by_region)
```

```python
# export the df_by_region df to csv file excluding index
df_by_region.to_csv('../data/cleaned/age22_29_ASFR_by_regional.csv', index=False)
```
